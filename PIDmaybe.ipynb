```python
#100 to 200
# t=data_100to200['Time'].reset_index(drop=True)/1000.0
# y_actual = data_100to200['Load Resistance'].reset_index(drop=True)

#200 to 500
t=data_200to500['Time'].reset_index(drop=True)/1000.0
y_actual = data_200to500['Load Resistance'].reset_index(drop=True)

y_start = y_actual[0]
y_target = 500
step_size = y_target - y_start
# 2. 1차 시스템(FOPDT) 함수 정의
def fopdt(t, K, tau, theta):

    response = np.zeros_like(t) # 결과 배열 초기화

    # 지연 시간(theta) 이후의 데이터만 계산
    mask = t >= theta

    if np.any(mask):
        exponent = -(t[mask] - theta) / tau
        # 지수값이 너무 작아지지 않게 클리핑 (선택사항, 안전용)
        # exponent = np.clip(exponent, -700, 700)

        val = K * (1 - np.exp(exponent))
        response[mask] = y_start + (step_size * val)

    # 지연 시간 이전은 시작값 그대로 유지
    response[~mask] = y_start

    return response

# 3. 최적의 파라미터(K, tau, theta) 찾기
# p0=[초기추측값] : Gain, Tau, Theta 순서
params, _ = curve_fit(fopdt, t, y_actual, p0=[100, 1, 0.5])
K_est, tau_est, theta_est = params
#theta_est = 0.001
print(f"Found system model -> Gain(K): {K_est:.2f}, 시상수(tau): {tau_est:.2f}, 지연(theta): {theta_est:.2f}")

# 4. 비교 그래프
plt.plot(t, y_actual, 'b.', label='Real Data')
plt.plot(t, fopdt(t, *params), 'r-', label='Fitted Model')
plt.legend()
plt.show()
```

```python
def calculate_pid_gains(K, tau, theta):
    """
    FOPDT 모델 파라미터(K, tau, theta)를 받아서
    다양한 튜닝 룰에 따른 PID 상수(Kp, Ki, Kd)를 반환합니다.
    """
    results = {}

    # ---------------------------------------------------------
    # 1. IMC (Internal Model Control) 튜닝 (가장 추천!)
    # 장점: 오버슈트(튀는 현상)를 조절하기 쉽습니다.
    # ---------------------------------------------------------

    # 조절 변수 Lambda (클수록 부드럽고, 작을수록 빠름)
    lambda_moderate = max(0.5 * tau, 2.0 * theta) # 보통 모드
    lambda_aggressive = max(0.1 * tau, 0.8 * theta) # 공격적 모드 (빠름)

    # [IMC - Moderate] 부드러운 제어
    Kp_imc_m = (1 / K) * (tau / (theta + lambda_moderate))
    Ti_imc_m = tau
    Td_imc_m = (tau * theta) / (2 * tau + theta)

    results['IMC'] = {
        'Kp': Kp_imc_m,
        'Ki': Kp_imc_m / Ti_imc_m,  # Ki = Kp / Ti
        'Kd': Kp_imc_m * Td_imc_m   # Kd = Kp * Td
    }

    # [IMC - Aggressive] 빠른 제어
    Kp_imc_a = (1 / K) * (tau / (theta + lambda_aggressive))
    Ti_imc_a = tau
    Td_imc_a = (tau * theta) / (2 * tau + theta)

    results['IMC_fast'] = {
        'Kp': Kp_imc_a,
        'Ki': Kp_imc_a / Ti_imc_a,
        'Kd': Kp_imc_a * Td_imc_a
    }

    # ---------------------------------------------------------
    # 2. Cohen-Coon 튜닝
    # 장점: 반응 속도가 매우 빠르지만, 약간 출렁거릴 수 있음.
    # ---------------------------------------------------------
    r = theta / tau  # 비율
    Kp_cc = (1/K) * (1.35/r + 0.27)
    Ti_cc = theta * (2.5*r + 0.5) / (2.5*r + 0.2)
    Td_cc = theta * (0.37*r) / (0.185*r + 0.1)

    results['Cohen-Coon'] = {
        'Kp': Kp_cc,
        'Ki': Kp_cc / Ti_cc,
        'Kd': Kp_cc * Td_cc
    }

    return results

# =========================================================
# 사용자 데이터 입력 (방금 구한 값)
# =========================================================
K_val = K_est
tau_val = tau_est
theta_val = theta_est

pid_candidates = calculate_pid_gains(K_val, tau_val, theta_val)

# 결과 출력
print(f"--- 분석 결과 (K={K_val}, tau={tau_val}, theta={theta_val}) ---")
for rule, gains in pid_candidates.items():
    print(f"\n[{rule}]")
    print(f"  P (Kp) : {gains['Kp']:.6f}")
    print(f"  I (Ki) : {gains['Ki']:.6f}")
    print(f"  D (Kd) : {gains['Kd']:.6f}")
```

```python
# ========================================================
# 1. 사용자 데이터 & 모델 설정 (여기만 맞춰주세요)
# ========================================================
# 아까 curve_fit 할 때 썼던 그 데이터 그대로 가져오세요 (인덱스 정리 안 해도 됨)
# 예시: t_raw = df['Time'].iloc[594:1000].values
#       y_raw = df['Load Resistance'].iloc[594:1000].values

# (가정: 사용자가 데이터를 넣었다고 치고 변수명만 맞춥니다)
# t_raw: 엄청 큰 숫자(Timestamp)들이 들어있는 배열
# y_actual: 실제 Load Resistance 값들

# 모델 파라미터 (아까 구한 값)
K = 1.01
tau = 3.55
theta = 0.77

# 시뮬레이션 해볼 PID 상수
Kp_sim, Ki_sim, Kd_sim = pid_candidates["IMC"]

# 목표값 (Set Resistance)
target_val = 200 # 100 -> 200 상황

# ========================================================
# 3. "불규칙한 dt"를 반영하는 정밀 시뮬레이터
# ========================================================
def simulate_real_system(time_array, y_start, setpoint, Kp, Ki, Kd, K, tau, theta):
    n = len(time_array)
    y_sim = np.zeros(n)
    u_history = np.zeros(n) # 제어 입력 기록 (지연 시간 처리용)

    # 초기값 설정
    y_sim[0] = y_start
    integral = 0
    prev_error = 0

    # 평균 dt 계산 (지연 시간 인덱스 추정용)
    avg_dt = np.mean(np.diff(time_array))
    delay_idx = int(theta / avg_dt) # 대략 몇 칸 전 데이터를 쓸지
    if delay_idx < 1: delay_idx = 1

    for i in range(1, n):
        # (1) 가변 dt 계산 (핵심!)
        dt = time_array[i] - time_array[i-1]
        if dt <= 0: dt = 1e-6 # 0으로 나누기 방지

        # (2) PID 계산
        error = setpoint - y_sim[i-1]
        integral += error * dt
        derivative = (error - prev_error) / dt

        # PID 출력 (제어 노력)
        # 여기서 u는 "변화시켜야 할 양"이 아니라 "목표 입력을 유지하려는 힘"
        u_val = (Kp * error) + (Ki * integral) + (Kd * derivative)

        # 실제 기계의 한계 반영 (Saturation, 예: PWM 범위 등)
        # u_val = np.clip(u_val, -1000, 1000)

        u_history[i] = u_val
        prev_error = error

        # (3) 시스템 반응 (FOPDT 이산화 모델 - Euler Method)
        # 지연된 입력(u) 가져오기
        idx_delayed = i - delay_idx
        if idx_delayed < 0:
            u_delayed = 0 # 시작 전엔 입력 없음
        else:
            u_delayed = u_history[idx_delayed]

        # 1차 지연 방정식: y[k] = (tau * y[k-1] + K * u_delayed * dt) / (tau + dt)
        # 이 식은 dt가 변해도 자동으로 보정됩니다.

        # 주의: u_delayed는 "변화량"을 의미하므로, y_sim은 "시작값 + 변화량" 구조여야 함.
        # 모델 수식의 안정성을 위해 변화분(dy)만 계산해서 더합니다.

        # 현재 값에서 목표값(K*u)으로 가려는 기울기
        # (K * u_delayed)는 최종적으로 도달할 변화량(Delta)
        target_delta = K * u_delayed
        current_delta = y_sim[i-1] - y_start

        # 1차 지연 미분방정식의 이산화: (목표 - 현재) / tau * dt
        change = (target_delta - current_delta) / tau * dt

        y_sim[i] = y_sim[i-1] + change

    return y_sim

# ========================================================
# 4. 실행 및 그래프
# ========================================================
y_simulated = simulate_real_system(
    t,
    y_actual[0], # 시작값 (100)
    target_val - y_actual[0], # PID에게는 "얼마나 더 변해야 하는지(Step Size)"를 줌
    Kp_sim, Ki_sim, Kd_sim,
    K, tau, theta
)

# 그래프 그리기
plt.figure(figsize=(12, 6))
plt.plot(t, y_actual, 'b-', alpha=0.3, label='Actual Data (Old PID)')
plt.plot(t, y_simulated, 'r-', linewidth=2, label='Simulation (New PID)')
plt.axhline(y=target_val, color='k', linestyle='--', label='Target')
plt.xlabel('Time (sec)')
plt.ylabel('Load Resistance')
plt.legend()
plt.grid(True)
plt.show()
```
